```python
import numpy as np
customers = np.array([
    [22, 2, 15, 1],
    [25, 4, 20, 2],
    [47, 10, 120, 12],
    [52, 12, 150, 15],
    [35, 6, 60, 6],
    [38, 7, 70, 7],
    [19, 1, 10, 1],
    [60, 8, 130, 10],
    [41, 9, 90, 9],
    [28, 5, 35, 4],
    [33, 6, 55, 5],
    [55, 11, 160, 14]
])
print(f"Type: {customers.dtype}")
print(f"Size: {customers.size}")
print(f"Shape: {customers.shape}")
print(f"Dimention: {customers.ndim}")
```

Rows show samples or data of each customer's information
Coulmns show the feature type, for example column one shows age, and column 2 shows monthly visit for each costumer.
This problem is an unsupervised learning, as we do not have the lables. This could be a clustring problem.

```python
mean_costume=np.mean(customers,axis=0)
std_costume=np.std(customers,axis=0)
max_costume=np.max(customers,axis=0)
min_costume=np.min(customers,axis=0)
print(f"Mean features: {mean_costume}")
print(f"Std features: {std_costume}")
print(f"Max features: {max_costume}")
print(f"Min features: {min_costume}")
feature_names = np.array([
 "age",
 "monthly_visits",
 "avg_order_value",
 "total_purchases"
])
print(feature_names)
for i in range(len(feature_names)):
    print(feature_names[i])
    print(f"i : mean= {mean_costume[0]},std= {std_costume[0]}, max= {max_costume[0]}, and min= {min_costume[0]}")
maximum_mean=np.argmax(mean_costume)
maximum_std=np.argmax(std_costume)
print(f"Maximum mean of data is {mean_costume[maximum_mean]} which is {feature_names[maximum_mean]}")
print(f"Maximum std of data is {std_costume[maximum_std]} which is {std_costume[maximum_std]}")
```

```python
customers_scaled=(customers-min_costume)/(max_costume-min_costume)
print(customers_scaled)
print(f" min: {np.min(customers_scaled)}")
print(f" max: {np.max(customers_scaled)}")
mean_customers_scaled=np.mean(customers_scaled,axis=0)    
print(f"Mean before scaling: {mean_costume}\nMean affter scaling: {mean_customers_scaled}")  
```

```python
ref_points = np.array([
    [23, 2, 15, 1], # low activity
    [32, 5, 45, 4], # normal
    [42, 8, 85, 8], # loyal
    [55, 12, 150, 14] # VIP
])
ref_points_scaled=(ref_points-min_costume)/(max_costume-min_costume)
print(ref_points_scaled)
e_distance=np.linalg.norm((customers_scaled[:,np.newaxis,:]-ref_points_scaled[np.newaxis,:,:]),axis=2)
print(e_distance)
```

```python
index=np.argmin(e_distance,axis=1)
print(index)
ref_names = np.array([
    "low activity",
     "normal",
    "loyal",
    "VIP"
])
for i in range(len(index)):
    print(f"Costumer {i+1} --> {ref_names[index[i]]}.")
```

```python
low_activity=np.where(index==0)
normal=np.where(index==1)
loyal=np.where(index==2)
VIP=np.where(index==3)
print(low_activity)
n_low=len(low_activity[0])
n_normal=len(normal[0])
n_loyal=len(loyal[0])
n_vip=len(VIP[0])
print(n_low,n_normal,n_loyal,n_vip )
len_clusters=np.vstack([n_low, n_normal, n_loyal, n_vip])
print(len_clusters)
```

```python
mean_low_activity=np.sum(customers[low_activity],axis=0)/n_low
mean_normal=np.sum(customers[normal],axis=0)/n_normal
mean_loyal=np.sum(customers[loyal],axis=0)/n_normal
mean_vip=np.sum(customers[VIP],axis=0)/n_normal
mean_features=np.vstack([mean_low_activity,mean_normal,mean_loyal,mean_vip])
print(mean_features)
```

```python
ref_names = np.array([
    "low activity",
     "normal",
    "loyal",
    "VIP"
])
for i in range(len(ref_names)):
    print(f"{ref_names[i]}")
    print(f"Counts: {len_clusters[i][0]}")
    print(f"MEan age: {mean_features[i,0]}")
    print(f"Mean monthly visits: {mean_features[i,1]}")
    print(f"Mean avg order value: {mean_features[i,2]}")
    print(f"Mean total purchases: {mean_features[i,3]}")
max_avg_order_value=np.argmax(mean_features[:,2])
print(f"The cluster of {ref_names[max_avg_order_value]} has highest average order value.")
max_monthly=np.argmax(mean_features[:,1])
print(f"The cluster of {ref_names[max_monthly]} has highest average monthly visit.")
    
```

```python
customer_label=np.where(customers[:,-1]>=8,"valuable","regular")
print(customer_label)
n_valuable=np.sum(customer_label=='valuable')
n_regular=np.sum(customer_label=='regular')
print(n_valuable,n_regular)
ind_v=customer_label=="valuable"
ind_r=customer_label=="regular"
print(ind_v)
print(ind_r)
mean_avrage_order_valuable=np.sum(customers[ind_v==True,2])/np.sum(ind_v)
mean_avrage_order_regular=np.sum(customers[ind_r==True,2])/np.sum(ind_r)
print(f"Average order of valuable customes: {mean_avrage_order_valuable}")
print(f"Average order of Regular customes: {np.round(mean_avrage_order_regular)}\n")

mean_monthly_visit_valuable=np.sum(customers[ind_v==True,1])/np.sum(ind_v)
mean_monthly_visit_regular=np.sum(customers[ind_r==True,1])/np.sum(ind_r)
print(f"Average Monthly visit of valuable customes: {mean_monthly_visit_valuable}")
print(f"Average Monthly visit of Regular customes: {np.round(mean_monthly_visit_regular)}\n")
```

```python
outlier_customer = np.array([[30, 1, 500, 1]])
customers_with_outlier=np.vstack([customers,outlier_customer])
# customers_with_outlier=np.concatenate((customers,outlier_customer),axis=0)
print(customers_with_outlier.shape)
scaled_customers_with_outlier=(customers_with_outlier-min_costume)/(max_costume-min_costume)
print(scaled_customers_with_outlier)
center = np.mean(scaled_customers_with_outlier, axis=0)
print(center,"\n")
distance_with_center=np.linalg.norm((scaled_customers_with_outlier-center),axis=1)
print(distance_with_center)
ind=np.argmax(distance_with_center)
print(ind)
```

داده پرت داده ای هست که هم از نظر میانگینی و هم از نظر انحراف معیار از مرکز خیلی دورتر باشد نسبت به سایر داده های موجود.
این مشتری داده پرت هست چون زمانی ک داده مقیاس بندی شد  سطر آخر و ستون سوم مقدار 1 را نشان نداد. همچنین مطابق با سایر ویزگی های ثبت شده داده ظاهرا این مقدار متوسط سفارش نمیتواند برای یک جامعه آماری بزرگ قابل تعمیم باشد. ویژگی ها بهم مرتبط هستن و روی مقدار هر کدام میتوانند موثر باشند.
در این تمرین برای داده پرت لیبل نداشتیم.
یک مسئله سوپرئایزد لرنینگ هست زیرا لیبل داده ها تعریف شده است.

```python
diff = customers_scaled[:, np.newaxis, :] - customers_scaled[np.newaxis, :, :]
E_distance_all_customers=np.linalg.norm(diff,axis=2)
# print(E_distance_all_customers)
diagnal_elements=np.diag(E_distance_all_customers)
print(diagnal_elements)
# The main diagnol should be 0 because distance between each elements with itself is 0.
np.fill_diagonal(E_distance_all_customers,np.inf)
# print(E_distance_all_customers)
ind=np.argmin(E_distance_all_customers,axis=1)
print(ind)
for i in range(len(ind)):
    print(f"Customer {i+1} nearest neighbor is Customer {ind[i]+1}")
```

```python
# Ground True
y_train = np.array(["pass", "fail", "pass", "pass", "fail", "pass"])
y_test = np.array(["pass", "fail", "fail", "pass"])
# Model A
model_A_train_pred = np.array(["pass", "fail", "pass", "pass", "fail", "pass"])
model_A_test_pred = np.array(["fail", "pass", "pass", "fail"])
# Model B
model_B_train_pred = np.array(["pass", "pass", "pass", "pass", "pass", "pass"])
model_B_test_pred = np.array(["pass", "pass", "pass", "pass"])
# Model C
model_C_train_pred = np.array(["pass", "fail", "pass", "fail", "fail", "pass"])
model_C_test_pred = np.array(["pass", "fail", "fail", "pass"])
# Train Accuracy
model_A_tr_accuracy=(len(np.where(model_A_train_pred==y_train)[0])/len(y_train))*100
model_B_tr_accuracy=(len(np.where(model_B_train_pred==y_train)[0])/len(y_train))*100
model_C_tr_accuracy=(len(np.where(model_C_train_pred==y_train)[0])/len(y_train))*100
# Test accuracy
model_A_ts_accuracy=(len(np.where(model_A_test_pred==y_test)[0])/len(y_test))*100
model_B_ts_accuracy=(len(np.where(model_B_test_pred==y_test)[0])/len(y_test))*100
model_C_ts_accuracy=(len(np.where(model_C_test_pred==y_test)[0])/len(y_test))*100
# Print accuracy of each model in train and test
print(f"Model A - Train Accuracy: {model_A_tr_accuracy} Test Accuracy:  {model_A_ts_accuracy}.")
print(f"Model A - Train Accuracy: {model_B_tr_accuracy} Test Accuracy:  {model_B_ts_accuracy}.")
print(f"Model A - Train Accuracy: {model_C_tr_accuracy} Test Accuracy:  {model_C_ts_accuracy}.")
```

```python
# Ground truth
y_train_reg = np.array([50, 75, 38, 92, 65, 84])
y_test_reg = np.array([55, 80, 60, 90])
# Model A
model_A_train_pred = np.array([50, 75, 38, 92, 65, 84])
model_A_test_pred = np.array([30, 100, 40, 120])
# Model B
model_B_train_pred = np.array([60, 60, 60, 60, 60, 60])
model_B_test_pred = np.array([60, 60, 60, 60])
# Model C
model_C_train_pred = np.array([52, 73, 40, 88, 66, 82])
model_C_test_pred = np.array([57, 78, 63, 88])
# Train mean abolute error
MAE_tr_A=np.mean(abs(model_A_train_pred-y_train_reg))
MAE_tr_B=np.mean(abs(model_B_train_pred-y_train_reg))
MAE_tr_C=np.mean(abs(model_C_train_pred-y_train_reg))
# Test mean abolute error
MAE_ts_A=np.mean(abs(model_A_test_pred-y_test_reg))
MAE_ts_B=np.mean(abs(model_B_test_pred-y_test_reg))
MAE_ts_C=np.mean(abs(model_C_test_pred-y_test_reg))
# Print accuracy of each model in train and test
print(f"Model A - Train MAE: {MAE_tr_A} Test MAE:  {MAE_ts_A}.")
print(f"Model A - Train MAE: {MAE_tr_B} Test MAE:  {MAE_ts_B}.")
print(f"Model A - Train MAE: {MAE_tr_C} Test MAE:  {MAE_ts_C}.")
```

```python
X = np.array([
 [2, 60, 45],
 [5, 80, 70],
 [1, 40, 35],
 [8, 90, 88],
 [4, 75, 60],
 [7, 85, 80]
])
y = np.array([50, 75, 38, 92, 65, 84])
prediction_simple = X[:, 0] * 10
abs_simple=np.abs(prediction_simple-y)
prediction_better = 0.4 * X[:, 0] * 10 + 0.3 * X[:, 1] + 0.3 * X[:, 2]
abs_better=np.abs(prediction_better-y)
MAE_simple=np.mean(np.abs(prediction_simple-y))
MAE_better=np.mean(np.abs(prediction_better-y))
print(MAE_simple,"\n",MAE_better)
```

```python
models = np.array(["Model A", "Model B", "Model C"])
train_error = np.array([2, 25, 8])
test_error = np.array([40, 30, 10])
diff_tr_ts=np.abs(train_error-test_error)
ind_overfitting=np.argmax(diff_tr_ts)
ind_generalizing=np.argmin(diff_tr_ts)
ind_underfitting=np.where((diff_tr_ts!=np.max(diff_tr_ts)) & (diff_tr_ts!=np.min(diff_tr_ts)))[0][0]
print(f"Model with overditting: {models[ind_overfitting]}")
print(f"Model with underfitting: {models[ind_underfitting]}")
print(f"Model with generalization: {models[ind_generalizing]}")
```
