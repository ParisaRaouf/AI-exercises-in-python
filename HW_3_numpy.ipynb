```python
import numpy as np
gym_data = np.array([
    [28, 75, 175, 4],
    [34, 68, 168, 3],
    [45, 82, 180, 2],
    [22, 58, 162, 5],
    [38, 90, 0, 1],
    [29, 65, 170, 0]
])
member_names = np.array(["Ali", "Sara", "Reza", "Neda", "Hassan",
"Maryam"])
## Section 1 of Question 1
data=gym_data[:,1:]
ind=data !=0
means=np.sum(data,axis=0)/np.sum(ind,axis=0)
row,columns=np.where(data==0)
gym_data[row,columns+1]=np.round(means[columns])
## Section 2 of Question 1
new_gym_data=gym_data[:,1:]
## Section 3 of Question 1
BMI=new_gym_data[:,0]/((new_gym_data[:,1]/100)**2)
physical_fitness_score=BMI+new_gym_data[:,2]*2
new_gym_data=np.column_stack((new_gym_data,physical_fitness_score))
## Section 4 of Question 1
maximum_ind=np.argmin(new_gym_data[-1])
print(f"Person with maximum physical fitness score: {member_names[maximum_ind]}")
print(new_gym_data)
## Section 5 of Question 1
member_mean=np.mean(new_gym_data[:,2])
member_S_D=np.std(new_gym_data[:,2])
distance_with_std=new_gym_data[:,2]-member_S_D
difference=np.abs(new_gym_data[:,2]-member_mean)/member_S_D
Max_diff=np.max(difference)
ind=np.where(difference==Max_diff)
names=member_names[ind]
print(difference)
print(f"These members have the most differences with others: {names}.")
```

```python
############ Section 1 of Question 2
recipes = np.array([
    [15, 350, 2, 5],
    [45, 600, 7, 10],
    [10, 200, 0, 3],
    [30, 450, 5, 7],
    [60, 800, 8, 12]
])
recipe_names = np.array(["Salad", "Curry", "Toast", "Pasta", "Stew"])
min_fea=np.min(recipes,axis=0)
max_fea=np.max(recipes,axis=0)
# print(min_fea)
# print(max_fea)
scaled_recipes=(recipes-min_fea)/(max_fea-min_fea)
# print(scaled_recipes)
########### Section 2 of Question 2
users = np.array([
    [10, 250, 1, 4],
    [50, 700, 8, 11],
    [25, 400, 4, 6]
])
scaled_users=(users-min_fea)/(max_fea-min_fea)
# print(scaled_users)
########### Section 3 of Question 2
diff=scaled_recipes[np.newaxis,:,:]-scaled_users[:,np.newaxis,:]
Euclidean_dis=np.linalg.norm(diff,axis=2)
# print(Euclidean_dis)
############ Section 4 of Question 2
food_taste=np.argmin(Euclidean_dis,axis=1)
print(food_taste)
foods=recipe_names[food_taste]
print(f"User one's food taste is most likely '{foods[0]}'.")
print(f"User two's food taste is most likely '{foods[1]}'.")
print(f"User three's food taste is most likely '{foods[2]}'.")
############ Section 5 of Question 2
foods_sorting=np.argsort(Euclidean_dis)
# print(foods_sorting)
print(f"User one's food tastes, repectively are '{recipe_names[foods_sorting[0,:]]}'.")
print(f"User two's food tastes,repectively are '{recipe_names[foods_sorting[1,:]]}'")
print(f"User three's food tastes, repectively are '{recipe_names[foods_sorting[2,:]]}'")
```

```python
############ Section 1 of Question 6
bib_numbers = np.array([101, 102, 103, 104, 105, 106])
times_5k = np.array([22.3, 25.1, 21.8, 26.4, 23.0, 24.7])
bib_numbers_2 = np.array([107, 108])
times_5k_2 = np.array([20.5, 27.9])
bib_numbers=np.concatenate((bib_numbers,bib_numbers_2),axis=0)
times_5k=np.concatenate((times_5k,times_5k_2),axis=0)
print(f"Member number: {bib_numbers}")
print(f"Member speed: {times_5k }\n")
############ Section 2 of Question 6
runner_data=np.vstack([bib_numbers, times_5k ])
np.random.seed(42)
indecies=np.random.permutation(len(runner_data[1,:]))
print(indecies)
shuffeled_data=runner_data[:,indecies]
print(runner_data,"\n")
print(f"shuffeled data:\n{shuffeled_data}")
############ Section 3 of Question 6
sorted_data_ind=np.argsort(times_5k)
print(sorted_data_ind)
counter=0
print("_______________________________________________________________________")
print(f"|   Runner's Number  |   Runner's Time Speed   |   Runner's Rank      |")
for i in range(len(sorted_data_ind)):
    print("_______________________________________________________________________")
    counter+=1
    print(f"|      {bib_numbers[sorted_data_ind[i]]}           |           {times_5k[sorted_data_ind[i]]}          |        {counter}             |")
print("_______________________________________________________________________")
############ Section 4 of Question 6
ind_104=np.where(bib_numbers==104)
print(ind_104)
rank=np.where(sorted_data_ind==ind_104[0][0])
print(f"Rank of runner number 104 is:\n{rank[0][0]+1} ")

```
